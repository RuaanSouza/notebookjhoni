## GERAÇÃO DE DADOS PARA OS DESAFIOS

Execute esta célula primeiro para gerar todos os arquivos de dados necessários para os desafios.


```python
import os
import json
import csv
import random
from datetime import datetime, timedelta

# Criar diretório para arquivos
if not os.path.exists('dados'):
    os.makedirs('dados')

# 1. Dados de sensores IoT em uma fábrica
sensores_data = []
for i in range(1000):
    timestamp = datetime.now() - timedelta(hours=random.randint(0, 720))
    sensor_id = f"SENSOR_{random.randint(1, 50):03d}"
    temperatura = round(random.normalvariate(25, 5), 2)
    umidade = round(random.normalvariate(60, 15), 2)
    pressao = round(random.normalvariate(1013, 20), 2)
    status = random.choice(['OK', 'ALERTA', 'CRITICO'])

    sensores_data.append({
        'timestamp': timestamp.strftime('%Y-%m-%d %H:%M:%S'),
        'sensor_id': sensor_id,
        'temperatura': temperatura,
        'umidade': umidade,
        'pressao': pressao,
        'status': status
    })

# Salvar dados de sensores
with open('dados/sensores_iot.json', 'w') as f:
    json.dump(sensores_data, f, indent=2)

# 2. Dados de produção industrial
producao_data = []
linhas_producao = ['LINHA_A', 'LINHA_B', 'LINHA_C', 'LINHA_D']
produtos = ['PRODUTO_X1', 'PRODUTO_X2', 'PRODUTO_X3', 'PRODUTO_Y1', 'PRODUTO_Y2']

for i in range(500):
    data = (datetime.now() - timedelta(days=random.randint(0, 90))).strftime('%Y-%m-%d')
    linha = random.choice(linhas_producao)
    produto = random.choice(produtos)
    quantidade = random.randint(100, 1000)
    defeitos = random.randint(0, 50)
    tempo_producao = round(random.normalvariate(8, 2), 2)

    producao_data.append([data, linha, produto, quantidade, defeitos, tempo_producao])

# Salvar dados de produção
with open('dados/producao_industrial.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['data', 'linha_producao', 'produto', 'quantidade', 'defeitos', 'tempo_producao_horas'])
    writer.writerows(producao_data)

# 3. Dados de qualidade de produtos
qualidade_data = []
for i in range(300):
    lote_id = f"LOTE_{random.randint(1000, 9999)}"
    peso = round(random.normalvariate(500, 50), 2)
    dimensao_x = round(random.normalvariate(10, 0.5), 2)
    dimensao_y = round(random.normalvariate(15, 0.8), 2)
    cor_score = round(random.uniform(0, 100), 2)
    dureza = round(random.normalvariate(75, 10), 2)
    aprovado = random.choice([True, False])

    qualidade_data.append({
        'lote_id': lote_id,
        'peso': peso,
        'dimensao_x': dimensao_x,
        'dimensao_y': dimensao_y,
        'cor_score': cor_score,
        'dureza': dureza,
        'aprovado': aprovado
    })

with open('dados/controle_qualidade.json', 'w') as f:
    json.dump(qualidade_data, f, indent=2)

# 4. Dados de manutenção preditiva
manutencao_data = []
equipamentos = ['MOTOR_001', 'MOTOR_002', 'BOMBA_001', 'BOMBA_002', 'COMPRESSOR_001']

for equip in equipamentos:
    for i in range(200):
        timestamp = (datetime.now() - timedelta(hours=random.randint(0, 2000))).strftime('%Y-%m-%d %H:%M:%S')
        vibracao = round(random.normalvariate(2.5, 1.0), 3)
        temperatura_motor = round(random.normalvariate(70, 15), 2)
        corrente = round(random.normalvariate(15, 3), 2)
        horas_operacao = random.randint(1000, 50000)
        falha_proximas_72h = random.choice([0, 1])

        manutencao_data.append([timestamp, equip, vibracao, temperatura_motor, corrente, horas_operacao, falha_proximas_72h])

with open('dados/manutencao_preditiva.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['timestamp', 'equipamento', 'vibracao', 'temperatura', 'corrente', 'horas_operacao', 'falha_72h'])
    writer.writerows(manutencao_data)

print("Dados gerados com sucesso!")
print("Arquivos criados:")
print("- dados/sensores_iot.json")
print("- dados/producao_industrial.csv")
print("- dados/controle_qualidade.json")
print("- dados/manutencao_preditiva.csv")
```

```python
import os
import json
import csv
import random
from datetime import datetime, timedelta
from collections import defaultdict

# ==============================================================================
# 1. SCRIPT ORIGINAL PARA GERAÇÃO DE DADOS (FORNECIDO PELO USUÁRIO)
# ==============================================================================
# Garante que o diretório de dados exista
if not os.path.exists('dados'):
    os.makedirs('dados')

# Gera o arquivo sensores_iot.json
sensores_data = []
for i in range(1000):
    timestamp = datetime.now() - timedelta(hours=random.randint(0, 720))
    sensor_id = f"SENSOR_{random.randint(1, 50):03d}"
    temperatura = round(random.normalvariate(25, 5), 2)
    umidade = round(random.normalvariate(60, 15), 2)
    pressao = round(random.normalvariate(1013, 20), 2)
    status = random.choice(['OK', 'ALERTA', 'CRITICO'])
    sensores_data.append({
        'timestamp': timestamp.strftime('%Y-%m-%d %H:%M:%S'), 'sensor_id': sensor_id,
        'temperatura': temperatura, 'umidade': umidade, 'pressao': pressao, 'status': status
    })
with open('dados/sensores_iot.json', 'w', encoding='utf-8') as f:
    json.dump(sensores_data, f, indent=2, ensure_ascii=False)

# Gera o arquivo producao_industrial.csv
producao_data = []
linhas_producao = ['LINHA_A', 'LINHA_B', 'LINHA_C', 'LINHA_D']
produtos = ['PRODUTO_X1', 'PRODUTO_X2', 'PRODUTO_X3', 'PRODUTO_Y1', 'PRODUTO_Y2']
for i in range(500):
    data = (datetime.now() - timedelta(days=random.randint(0, 90))).strftime('%Y-%m-%d')
    linha = random.choice(linhas_producao)
    produto = random.choice(produtos)
    quantidade = random.randint(100, 1000)
    defeitos = random.randint(0, 50)
    tempo_producao = round(random.normalvariate(8, 2), 2)
    producao_data.append([data, linha, produto, quantidade, defeitos, tempo_producao])
with open('dados/producao_industrial.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['data', 'linha_producao', 'produto', 'quantidade', 'defeitos', 'tempo_producao_horas'])
    writer.writerows(producao_data)

# Gera o arquivo controle_qualidade.json
qualidade_data = []
for i in range(300):
    lote_id = f"LOTE_{random.randint(1000, 9999)}"
    peso = round(random.normalvariate(500, 50), 2)
    dimensao_x = round(random.normalvariate(10, 0.5), 2)
    dimensao_y = round(random.normalvariate(15, 0.8), 2)
    cor_score = round(random.uniform(0, 100), 2)
    dureza = round(random.normalvariate(75, 10), 2)
    aprovado = random.choice([True, False])
    qualidade_data.append({
        'lote_id': lote_id, 'peso': peso, 'dimensao_x': dimensao_x, 'dimensao_y': dimensao_y,
        'cor_score': cor_score, 'dureza': dureza, 'aprovado': aprovado
    })
with open('dados/controle_qualidade.json', 'w', encoding='utf-8') as f:
    json.dump(qualidade_data, f, indent=2, ensure_ascii=False)

# Gera o arquivo manutencao_preditiva.csv
manutencao_data = []
equipamentos = ['MOTOR_001', 'MOTOR_002', 'BOMBA_001', 'BOMBA_002', 'COMPRESSOR_001']
for equip in equipamentos:
    for i in range(200):
        timestamp = (datetime.now() - timedelta(hours=random.randint(0, 2000))).strftime('%Y-%m-%d %H:%M:%S')
        vibracao = round(random.normalvariate(2.5, 1.0), 3)
        temperatura_motor = round(random.normalvariate(70, 15), 2)
        corrente = round(random.normalvariate(15, 3), 2)
        horas_operacao = random.randint(1000, 50000)
        falha_proximas_72h = random.choice([0, 1])
        manutencao_data.append([timestamp, equip, vibracao, temperatura_motor, corrente, horas_operacao, falha_proximas_72h])
with open('dados/manutencao_preditiva.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['timestamp', 'equipamento', 'vibracao', 'temperatura', 'corrente', 'horas_operacao', 'falha_72h'])
    writer.writerows(manutencao_data)

# Cria um arquivo de texto para os desafios 10 e 30
with open('dados/exemplo_texto.txt', 'w', encoding='utf-8') as f:
    f.write("A indústria 4.0 representa a automação industrial e a integração de diferentes tecnologias.\n")
    f.write("Análise de dados é fundamental para a manutenção preditiva e controle de qualidade.")

print("="*50)
print("Dados gerados e prontos para análise.")
print("="*50 + "\n")


# ==============================================================================
# 2. FUNÇÕES PARA CARREGAR OS DADOS GERADOS
# ==============================================================================

def carregar_sensores_iot():
    with open('dados/sensores_iot.json', 'r', encoding='utf-8') as f:
        return json.load(f)

def carregar_producao_industrial():
    with open('dados/producao_industrial.csv', 'r', newline='', encoding='utf-8') as f:
        return list(csv.DictReader(f))

def carregar_controle_qualidade():
    with open('dados/controle_qualidade.json', 'r', encoding='utf-8') as f:
        return json.load(f)

def carregar_manutencao_preditiva():
    with open('dados/manutencao_preditiva.csv', 'r', newline='', encoding='utf-8') as f:
        return list(csv.DictReader(f))

# Carrega todos os dados de uma vez
dados_sensores = carregar_sensores_iot()
dados_producao = carregar_producao_industrial()
dados_qualidade = carregar_controle_qualidade()
dados_manutencao = carregar_manutencao_preditiva()

# ==============================================================================
# 3. RESOLUÇÃO DOS 30 DESAFIOS
# ==============================================================================

# --- Desafio 1 ---
def resolver_desafio_1():
    print("--- Desafio 1: Classificar Sensor ---")
    def classificar_sensor(temperatura, umidade, pressao):
        if temperatura < 0 or umidade < 10 or pressao < 900: return "CRÍTICO"
        elif temperatura < 20 or umidade < 30 or pressao < 950: return "ALERTA"
        else: return "OK"

    # Testando com os 3 primeiros sensores do arquivo JSON
    for sensor in dados_sensores[:3]:
        classificacao = classificar_sensor(sensor['temperatura'], sensor['umidade'], sensor['pressao'])
        print(f"Sensor {sensor['sensor_id']} com T={sensor['temperatura']}°C, U={sensor['umidade']}%, P={sensor['pressao']}hPa -> Status: {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 2 ---
def resolver_desafio_2():
    print("--- Desafio 2: Classificar Produto por Categoria e Preço ---")
    def classificar_produto(categoria, preco):
        match categoria:
            case "X":
                if preco > 800: return "Produto Premium"
                elif preco > 400: return "Produto Médio"
                else: return "Produto Econômico"
            case "Y":
                if preco > 900: return "Produto Premium"
                elif preco > 450: return "Produto Médio"
                else: return "Produto Econômico"
            case _: return "Categoria desconhecida"

    # Nota: Preço e categoria não existem. Vamos simulá-los.
    # Categoria será 'X' ou 'Y' do nome do produto. Preço será simulado.
    for reg in dados_producao[:3]:
        categoria = reg['produto'][-2] # Pega 'X' ou 'Y'
        # CORREÇÃO APLICADA AQUI: Convertendo reg['quantidade'] para float
        preco_simulado = float(reg['quantidade']) * 1.5 # Simula um preço
        classificacao = classificar_produto(categoria, preco_simulado)
        print(f"Produto {reg['produto']} com preço simulado de R${preco_simulado:.2f} -> Classificação: {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 3 ---
def resolver_desafio_3():
    print("--- Desafio 3: Analisar Produção ---")
    def analisar_producao(producao_diaria):
        media_producao = sum(producao_diaria) / len(producao_diaria)
        max_producao = max(producao_diaria)
        min_producao = min(producao_diaria)
        return {"média": media_producao, "máximo": max_producao, "mínimo": min_producao}

    quantidades = [int(p['quantidade']) for p in dados_producao]
    analise = analisar_producao(quantidades)
    print(f"Análise de todas as quantidades produzidas: {analise}")
    print("-" * 20 + "\n")

# --- Desafio 4 ---
def resolver_desafio_4():
    print("--- Desafio 4: Filtrar Máquinas por Estado ---")
    # A função original não se aplica diretamente. Vamos adaptá-la.
    # Usaremos 'falha_72h' (0=OPERACIONAL, 1=FALHA_PREVISTA) como "estado".
    def filtrar_maquinas_por_estado(maquinas, estado_desejado):
        return [maq for maq in maquinas if int(maq['falha_72h']) == estado_desejado]

    maquinas_com_falha_prevista = filtrar_maquinas_por_estado(dados_manutencao, 1)
    print(f"Encontradas {len(maquinas_com_falha_prevista)} máquinas com previsão de falha.")
    # Imprime as 3 primeiras para exemplo
    for maq in maquinas_com_falha_prevista[:3]:
        print(f" - Equipamento: {maq['equipamento']} em {maq['timestamp']}")
    print("-" * 20 + "\n")

# --- Desafio 5 ---
def resolver_desafio_5():
    print("--- Desafio 5: Analisar Média de Coluna em CSV ---")
    def analisar_csv(caminho_arquivo, nome_coluna):
        with open(caminho_arquivo, newline='', encoding='utf-8') as csvfile:
            leitor = csv.DictReader(csvfile)
            valores = [float(linha[nome_coluna]) for linha in leitor if linha[nome_coluna]]
            return sum(valores) / len(valores) if valores else 0

    media_defeitos = analisar_csv('dados/producao_industrial.csv', 'defeitos')
    print(f"A média de defeitos por registro de produção é: {media_defeitos:.2f}")
    print("-" * 20 + "\n")

# --- Desafio 6 ---
def resolver_desafio_6():
    print("--- Desafio 6: Analisar Máquinas Produtivas ---")
    # O conceito de "produtividade > 75" é abstrato. Vamos adaptá-lo para
    # "horas de operação > 40000", um sinal de uma máquina muito utilizada.
    def analisar_maquinas_produtivas(maquinas, min_horas):
        maquinas_produtivas = [maq for maq in maquinas if int(maq['horas_operacao']) > min_horas]
        return maquinas_produtivas

    maquinas_muito_usadas = analisar_maquinas_produtivas(dados_manutencao, 40000)
    print(f"Encontrados {len(maquinas_muito_usadas)} registros de máquinas com mais de 40.000 horas de operação.")
    for reg in maquinas_muito_usadas[:3]:
        print(f" - Equipamento {reg['equipamento']} com {reg['horas_operacao']} horas.")
    print("-" * 20 + "\n")

# --- Desafio 7 ---
def resolver_desafio_7():
    print("--- Desafio 7: Transformar Produção (Acima da Média) ---")
    def transformar_producao(producao_diaria):
        if not producao_diaria: return []
        media = sum(producao_diaria) / len(producao_diaria)
        return [producao for producao in producao_diaria if producao > media]

    quantidades = [int(p['quantidade']) for p in dados_producao]
    acima_da_media = transformar_producao(quantidades)
    print(f"De {len(quantidades)} registros de produção, {len(acima_da_media)} estão acima da média.")
    print("-" * 20 + "\n")

# --- Desafio 8 ---
def resolver_desafio_8():
    print("--- Desafio 8: Analisar Média de Chave em JSON ---")
    def analisar_json(caminho_arquivo, chave):
        with open(caminho_arquivo, 'r', encoding='utf-8') as jsonfile:
            dados = json.load(jsonfile)
            valores = [item[chave] for item in dados if chave in item]
            return sum(valores) / len(valores) if valores else 0

    media_temperatura = analisar_json('dados/sensores_iot.json', 'temperatura')
    print(f"A média de temperatura de todos os sensores é: {media_temperatura:.2f}°C")
    print("-" * 20 + "\n")

# --- Desafio 9 ---
def resolver_desafio_9():
    print("--- Desafio 9: Classificar Produto por Preço e Quantidade ---")
    def classificar_produto_preco_quantidade(preco, quantidade):
        if preco > 1000 and quantidade > 500: return "Produto Premium em Grande Quantidade"
        elif preco > 500 and quantidade > 200: return "Produto Médio em Quantidade Moderada"
        elif preco <= 500 and quantidade <= 200: return "Produto Econômico com Estoque Baixo"
        else: return "Produto Médio ou Econômico"

    # Novamente, simulando o preço
    for reg in dados_producao[:3]:
        preco_simulado = float(reg['tempo_producao_horas']) * 50
        quantidade = int(reg['quantidade'])
        classificacao = classificar_produto_preco_quantidade(preco_simulado, quantidade)
        print(f"Produto com Qtd={quantidade} e Preço Simulado={preco_simulado:.2f} -> {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 10 ---
def resolver_desafio_10():
    print("--- Desafio 10: Contar Palavras em Arquivo ---")
    def contar_palavras_arquivo(caminho_arquivo):
        with open(caminho_arquivo, 'r', encoding='utf-8') as arquivo:
            return len(arquivo.read().split())

    total_palavras = contar_palavras_arquivo('dados/exemplo_texto.txt')
    print(f"O arquivo 'exemplo_texto.txt' contém {total_palavras} palavras.")
    print("-" * 20 + "\n")

# --- Desafio 11 ---
def resolver_desafio_11():
    print("--- Desafio 11: Sistema de Alertas Inteligente ---")
    # Adaptando a função para usar 'sensor_id' em vez de 'id'
    def sistema_alertas_inteligente(dados_sensores):
        alertas = []
        for sensor in dados_sensores:
            if sensor['temperatura'] > 35:
                alertas.append(f"Alerta! Temperatura ALTA no sensor {sensor['sensor_id']}")
            elif sensor['umidade'] < 20:
                alertas.append(f"Alerta! Umidade BAIXA no sensor {sensor['sensor_id']}")
            elif sensor['pressao'] > 1040:
                alertas.append(f"Alerta! Pressão ALTA no sensor {sensor['sensor_id']}")
        return alertas if alertas else ["Nenhum alerta crítico gerado."]

    alertas = sistema_alertas_inteligente(dados_sensores)
    print("Primeiros 5 alertas gerados:")
    for alerta in alertas[:5]:
        print(f" - {alerta}")
    print("-" * 20 + "\n")

# --- Desafio 12 ---
def resolver_desafio_12():
    print("--- Desafio 12: Simular Linha de Produção ---")
    def simular_linha_producao(dias_trabalho, producao_diaria_media):
        return dias_trabalho * producao_diaria_media

    # Calculando produção diária média a partir dos dados reais
    quantidades = [int(p['quantidade']) for p in dados_producao]
    dias_unicos = len(set(p['data'] for p in dados_producao))
    producao_media_por_registro = sum(quantidades) / len(quantidades) if quantidades else 0

    simulacao = simular_linha_producao(30, producao_media_por_registro)
    print(f"Usando uma produção média de {producao_media_por_registro:.2f} por registro/turno, a produção em 30 dias seria de: {simulacao:.0f} unidades.")
    print("-" * 20 + "\n")

# --- Desafio 13 ---
def resolver_desafio_13():
    print("--- Desafio 13: Dias de Maior Produção ---")
    # A função original espera uma lista. Vamos agrupar os dados por data primeiro.
    def dias_maior_producao(producao_por_data):
        if not producao_por_data: return []
        max_producao = max(producao_por_data.values())
        return [data for data, prod in producao_por_data.items() if prod == max_producao]

    producao_agrupada = defaultdict(int)
    for reg in dados_producao:
        producao_agrupada[reg['data']] += int(reg['quantidade'])

    dias_pico = dias_maior_producao(producao_agrupada)
    if dias_pico:
        pico_producao = producao_agrupada[dias_pico[0]]
        print(f"O pico de produção foi de {pico_producao} unidades, e ocorreu no(s) dia(s): {dias_pico}")
    print("-" * 20 + "\n")

# --- Desafio 14 ---
def resolver_desafio_14():
    print("--- Desafio 14: Classificar Sensor por Temperatura e Umidade ---")
    def classificar_sensor_temperatura_umidade(temperatura, umidade):
        if temperatura > 30 and umidade < 20: return "Sensor Crítico"
        elif temperatura > 25 and umidade < 40: return "Sensor de Alerta"
        else: return "Sensor OK"

    for sensor in dados_sensores[:3]:
        classificacao = classificar_sensor_temperatura_umidade(sensor['temperatura'], sensor['umidade'])
        print(f"Sensor {sensor['sensor_id']} (T={sensor['temperatura']}, U={sensor['umidade']}) -> Status: {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 15 ---
def resolver_desafio_15():
    print("--- Desafio 15: Calcular Eficiência de Produção ---")
    def calcular_eficiencia_producao(total_producao, dias_trabalho):
        return total_producao / dias_trabalho if dias_trabalho > 0 else 0

    total_producao = sum(int(p['quantidade']) for p in dados_producao)
    dias_trabalho = len(set(p['data'] for p in dados_producao))

    eficiencia = calcular_eficiencia_producao(total_producao, dias_trabalho)
    print(f"Total produzido: {total_producao} em {dias_trabalho} dias.")
    print(f"Eficiência média: {eficiencia:.2f} unidades por dia.")
    print("-" * 20 + "\n")

# --- Desafio 16 ---
def resolver_desafio_16():
    print("--- Desafio 16: Detectar Anomalias em Tempo Real (com Desvio Padrão) ---")
    def detectar_anomalias_tempo_real(fluxo_dados):
        if len(fluxo_dados) < 2: return []
        media = sum(fluxo_dados) / len(fluxo_dados)
        desvio_padrao = (sum([(x - media) ** 2 for x in fluxo_dados]) / len(fluxo_dados)) ** 0.5
        limiar_superior = media + 2 * desvio_padrao
        limiar_inferior = media - 2 * desvio_padrao
        return [valor for valor in fluxo_dados if valor > limiar_superior or valor < limiar_inferior]

    # Aplicando aos dados de vibração dos motores
    dados_vibracao = [float(d['vibracao']) for d in dados_manutencao]
    anomalias = detectar_anomalias_tempo_real(dados_vibracao)
    print(f"Detectadas {len(anomalias)} anomalias de vibração em {len(dados_vibracao)} medições.")
    print(f"Exemplos de anomalias: {anomalias[:5]}")
    print("-" * 20 + "\n")

# --- Desafio 17 ---
def resolver_desafio_17():
    print("--- Desafio 17: Classificar Produtos por Categoria (com Match-Case) ---")
    def classificar_produtos_categoria(categoria, preco):
        if categoria == "X":
            if preco > 800: return "Produto X Premium"
            else: return "Produto X Econômico"
        elif categoria == "Y":
            if preco > 900: return "Produto Y Premium"
            else: return "Produto Y Econômico"
        else: return "Categoria Desconhecida"

    # Similar ao desafio 2, com dados simulados
    for reg in dados_producao[:3]:
        categoria = reg['produto'][-2]
        # CORREÇÃO APLICADA AQUI: Convertendo reg['quantidade'] para float
        preco_simulado = float(reg['quantidade']) * 1.5
        classificacao = classificar_produtos_categoria(categoria, preco_simulado)
        print(f"Produto {reg['produto']} (preço simulado R${preco_simulado:.2f}) -> {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 18 ---
def resolver_desafio_18():
    print("--- Desafio 18: Calcular Média de Produção Diária (com Dicionário) ---")
    def calcular_media_producao_diaria(producao_diaria_dict):
        return sum(producao_diaria_dict.values()) / len(producao_diaria_dict) if producao_diaria_dict else 0

    producao_agrupada = defaultdict(int)
    for reg in dados_producao:
        producao_agrupada[reg['data']] += int(reg['quantidade'])

    media = calcular_media_producao_diaria(producao_agrupada)
    print(f"A média de produção por dia (considerando {len(producao_agrupada)} dias) é: {media:.2f} unidades.")
    print("-" * 20 + "\n")

# --- Desafio 19 ---
def resolver_desafio_19():
    print("--- Desafio 19: Detectar Anomalias com Tolerância Fixa ---")
    def detectar_anomalias_com_tolerancia(dados, tolerancia):
        if not dados: return []
        media = sum(dados) / len(dados)
        limiar_superior = media + tolerancia
        limiar_inferior = media - tolerancia
        return [valor for valor in dados if valor > limiar_superior or valor < limiar_inferior]

    # Aplicando ao peso dos lotes no controle de qualidade
    pesos_lotes = [lote['peso'] for lote in dados_qualidade]
    anomalias = detectar_anomalias_com_tolerancia(pesos_lotes, 40) # Tolerância de 40g
    print(f"Detectadas {len(anomalias)} anomalias de peso (tolerância de 40g) em {len(pesos_lotes)} lotes.")
    print(f"Exemplos de anomalias: {anomalias[:5]}")
    print("-" * 20 + "\n")

# --- Desafio 20 ---
def resolver_desafio_20():
    print("--- Desafio 20: Calcular Variação da Produção ---")
    def calcular_variacao_producao(producao_diaria):
        return max(producao_diaria) - min(producao_diaria) if producao_diaria else 0

    quantidades = [int(p['quantidade']) for p in dados_producao]
    variacao = calcular_variacao_producao(quantidades)
    print(f"A variação (diferença entre max e min) na quantidade produzida por registro é: {variacao}")
    print("-" * 20 + "\n")

# --- Desafio 21 (Duplicado do 16) ---
def resolver_desafio_21():
    print("--- Desafio 21: Detectar Anomalias em Tempo Real (Repetido) ---")
    def detectar_anomalias_tempo_real(fluxo_dados):
        if len(fluxo_dados) < 2: return []
        media = sum(fluxo_dados) / len(fluxo_dados)
        desvio_padrao = (sum([(x - media) ** 2 for x in fluxo_dados]) / len(fluxo_dados)) ** 0.5
        limiar_superior = media + 2 * desvio_padrao
        limiar_inferior = media - 2 * desvio_padrao
        return [valor for valor in fluxo_dados if valor > limiar_superior or valor < limiar_inferior]

    # Aplicando aos dados de corrente elétrica dos equipamentos
    dados_corrente = [float(d['corrente']) for d in dados_manutencao]
    anomalias = detectar_anomalias_tempo_real(dados_corrente)
    print(f"Detectadas {len(anomalias)} anomalias de corrente elétrica em {len(dados_corrente)} medições.")
    print(f"Exemplos de anomalias: {anomalias[:5]}")
    print("-" * 20 + "\n")

# --- Desafio 22 (Duplicado do 12) ---
def resolver_desafio_22():
    print("--- Desafio 22: Simular Linha de Produção (Repetido) ---")
    def simular_linha_producao(dias_trabalho, producao_diaria):
        return dias_trabalho * producao_diaria

    # Simulação para uma linha específica
    producao_linha_a = [int(p['quantidade']) for p in dados_producao if p['linha_producao'] == 'LINHA_A']
    media_linha_a = sum(producao_linha_a) / len(producao_linha_a) if producao_linha_a else 0
    simulacao = simular_linha_producao(30, media_linha_a)
    print(f"A LINHA_A produz em média {media_linha_a:.2f} por registro.")
    print(f"Simulação para 30 dias na LINHA_A: {simulacao:.0f} unidades.")
    print("-" * 20 + "\n")

# --- Desafio 23 (Duplicado do 20) ---
def resolver_desafio_23():
    print("--- Desafio 23: Calcular Variação da Produção (Repetido) ---")
    def calcular_variacao_producao(producao_diaria):
        return max(producao_diaria) - min(producao_diaria) if producao_diaria else 0

    # Aplicando à produção diária agregada
    producao_agrupada = defaultdict(int)
    for reg in dados_producao:
        producao_agrupada[reg['data']] += int(reg['quantidade'])

    valores_diarios = list(producao_agrupada.values())
    variacao = calcular_variacao_producao(valores_diarios)
    print(f"A variação na produção diária total (diferença entre o melhor e o pior dia) é: {variacao}")
    print("-" * 20 + "\n")

# --- Desafio 24 ---
def resolver_desafio_24():
    print("--- Desafio 24: Calcular Média de Temperaturas (com Tuplas) ---")
    def calcular_media_temperaturas(sensores_tuplas):
        return sum(sensor[1] for sensor in sensores_tuplas) / len(sensores_tuplas) if sensores_tuplas else 0

    # Convertendo os dados do JSON para o formato de tupla esperado
    sensores_formato_tupla = [(s['sensor_id'], s['temperatura']) for s in dados_sensores]
    media = calcular_media_temperaturas(sensores_formato_tupla)
    print(f"A média de temperatura (calculada a partir de uma lista de tuplas) é: {media:.2f}°C")
    print("-" * 20 + "\n")

# --- Desafio 25 (Duplicado do 18) ---
def resolver_desafio_25():
    print("--- Desafio 25: Calcular Média de Produção Diária (Repetido) ---")
    def calcular_media_producao_diaria(producao_diaria_dict):
        return sum(producao_diaria_dict.values()) / len(producao_diaria_dict) if producao_diaria_dict else 0

    producao_agrupada = defaultdict(int)
    for reg in dados_producao:
        producao_agrupada[reg['data']] += int(reg['quantidade'])

    media = calcular_media_producao_diaria(producao_agrupada)
    print(f"A média de produção por dia é: {media:.2f} unidades (repetição do desafio 18).")
    print("-" * 20 + "\n")

# --- Desafio 26 (Duplicado do 9) ---
def resolver_desafio_26():
    print("--- Desafio 26: Classificar Produto por Preço e Quantidade (Repetido) ---")
    def classificar_produto_preco_quantidade(preco, quantidade):
        if preco > 1000 and quantidade > 500: return "Produto Premium em Grande Quantidade"
        elif preco > 500 and quantidade > 200: return "Produto Médio em Quantidade Moderada"
        elif preco <= 500 and quantidade <= 200: return "Produto Econômico com Estoque Baixo"
        else: return "Produto Médio ou Econômico"

    # Testando com um dos últimos registros
    reg = dados_producao[-1]
    preco_simulado = float(reg['tempo_producao_horas']) * 50
    quantidade = int(reg['quantidade'])
    classificacao = classificar_produto_preco_quantidade(preco_simulado, quantidade)
    print(f"Produto com Qtd={quantidade} e Preço Simulado={preco_simulado:.2f} -> {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 27 ---
def resolver_desafio_27():
    print("--- Desafio 27: Classificar Temperatura de Sensores (com Tuplas) ---")
    def classificar_temperatura_sensores(sensores_tuplas):
        classificacoes = []
        for sensor in sensores_tuplas:
            if sensor[1] > 30: classificacoes.append(f"Sensor {sensor[0]}: Temperatura Alta")
            elif sensor[1] < 10: classificacoes.append(f"Sensor {sensor[0]}: Temperatura Baixa")
            else: classificacoes.append(f"Sensor {sensor[0]}: Temperatura Normal")
        return classificacoes

    # Convertendo os dados e aplicando a função
    sensores_formato_tupla = [(s['sensor_id'], s['temperatura']) for s in dados_sensores]
    classificacoes = classificar_temperatura_sensores(sensores_formato_tupla[:5]) # Apenas os 5 primeiros
    for c in classificacoes:
        print(f" - {c}")
    print("-" * 20 + "\n")

# --- Desafio 28 (Duplicado do 14) ---
def resolver_desafio_28():
    print("--- Desafio 28: Classificar Sensor por Temp e Umidade (Repetido) ---")
    def classificar_sensor_temperatura_umidade(temperatura, umidade):
        if temperatura > 30 and umidade < 20: return "Sensor Crítico"
        elif temperatura > 25 and umidade < 40: return "Sensor de Alerta"
        else: return "Sensor OK"

    for sensor in random.sample(dados_sensores, 3): # Pega 3 aleatórios
        classificacao = classificar_sensor_temperatura_umidade(sensor['temperatura'], sensor['umidade'])
        print(f"Sensor aleatório {sensor['sensor_id']} (T={sensor['temperatura']}, U={sensor['umidade']}) -> Status: {classificacao}")
    print("-" * 20 + "\n")

# --- Desafio 29 (Duplicado do 13) ---
def resolver_desafio_29():
    print("--- Desafio 29: Dias de Maior Produção (Repetido) ---")
    def dias_maior_producao(producao_por_data_list): # Função original espera uma lista
        if not producao_por_data_list: return []
        max_producao = max(producao_por_data_list)
        return [i + 1 for i, producao in enumerate(producao_por_data_list) if producao == max_producao]

    producao_agrupada = defaultdict(int)
    for reg in dados_producao:
        producao_agrupada[reg['data']] += int(reg['quantidade'])

    # Ordena por data para que o índice tenha sentido sequencial
    producao_ordenada = sorted(producao_agrupada.items())
    lista_producao = [prod for data, prod in producao_ordenada]

    indices_pico = dias_maior_producao(lista_producao)
    print(f"A produção máxima ocorreu no(s) dia(s) de índice (sequencial): {indices_pico}")
    for i in indices_pico:
        data_pico = producao_ordenada[i-1][0]
        valor_pico = producao_ordenada[i-1][1]
        print(f" - Índice {i} corresponde à data {data_pico} com {valor_pico} unidades.")
    print("-" * 20 + "\n")

# --- Desafio 30 (Duplicado do 10) ---
def resolver_desafio_30():
    print("--- Desafio 30: Contar Palavras em Arquivo (Repetido) ---")
    def contar_palavras_arquivo(caminho_arquivo):
        with open(caminho_arquivo, 'r', encoding='utf-8') as arquivo:
            return len(arquivo.read().split())

    total_palavras = contar_palavras_arquivo('dados/exemplo_texto.txt')
    print(f"O arquivo 'exemplo_texto.txt' contém {total_palavras} palavras.")
    print("-" * 20 + "\n")

# ==============================================================================
# 4. BLOCO DE EXECUÇÃO PRINCIPAL
# ==============================================================================
if __name__ == "__main__":
    # Executa todos os desafios em ordem
    resolver_desafio_1()
    resolver_desafio_2()
    resolver_desafio_3()
    resolver_desafio_4()
    resolver_desafio_5()
    resolver_desafio_6()
    resolver_desafio_7()
    resolver_desafio_8()
    resolver_desafio_9()
    resolver_desafio_10()
    resolver_desafio_11()
    resolver_desafio_12()
    resolver_desafio_13()
    resolver_desafio_14()
    resolver_desafio_15()
    resolver_desafio_16()
    resolver_desafio_17()
    resolver_desafio_18()
    resolver_desafio_19()
    resolver_desafio_20()
    resolver_desafio_21()
    resolver_desafio_22()
    resolver_desafio_23()
    resolver_desafio_24()
    resolver_desafio_25()
    resolver_desafio_26()
    resolver_desafio_27()
    resolver_desafio_28()
    resolver_desafio_29()
    resolver_desafio_30()
```
